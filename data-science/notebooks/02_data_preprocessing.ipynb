```python
target = "Congestion Level"

features = [
    "Area Name",
    "Road/Intersection Name",
    "Traffic Volume",
    "Average Speed",
    "Travel Time Index",
    "Road Capacity Utilization",
    "Incident Reports",
    "Public Transport Usage",
    "Traffic Signal Compliance",
    "Parking Usage",
    "Pedestrian and Cyclist Count",
    "Weather Conditions",
    "Roadwork and Construction Activity",
    "Year",
    "Month",
    "Day",
    "Day_of_Week"
]

X = df[features]
y = df[target]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print(y.name)
```

```python
categorical_features = [
    "Area Name",
    "Road/Intersection Name",
    "Weather Conditions",
    "Roadwork and Construction Activity",
    "Day_of_Week"
]

numerical_features = [
    "Traffic Volume",
    "Average Speed",
    "Travel Time Index",
    "Road Capacity Utilization",
    "Incident Reports",
    "Public Transport Usage",
    "Traffic Signal Compliance",
    "Parking Usage",
    "Pedestrian and Cyclist Count",
    "Year",
    "Month",
    "Day"
]

print("Categorical features:", len(categorical_features))
print(categorical_features)

print("\nNumerical features:", len(numerical_features))
print(numerical_features)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("num", StandardScaler(), numerical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("X_train processed shape:", X_train_processed.shape)
print("X_test processed shape:", X_test_processed.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train_processed, y_train)

print("Random Forest model trained successfully!")
```

```python
y_pred = model.predict(X_test_processed)

print("Prediction completed successfully!")
print("Number of predictions:", len(y_pred))

print("\nFirst 10 predictions:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(mean_squared_error(y_test, y_pred))

r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")
print("------------------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R² Score:", r2)
```

```python
comparison = pd.DataFrame({
    "Actual Congestion": y_test.values,
    "Predicted Congestion": y_pred
})

comparison["Error"] = (
    comparison["Actual Congestion"]
    - comparison["Predicted Congestion"]
)

print(comparison.head(10))
```

```python
# Recreate the processed data for the 15-feature model

X_check = df[features_no_capacity_tti]
y_check = df["Congestion Level"]

X_train_check, X_test_check, y_train_check, y_test_check = train_test_split(
    X_check,
    y_check,
    test_size=0.20,
    random_state=42
)

X_train_check_processed = preprocessor_check.fit_transform(X_train_check)
X_test_check_processed = preprocessor_check.transform(X_test_check)

print("Processed training shape:", X_train_check_processed.shape)
print("Processed testing shape:", X_test_check_processed.shape)
```

```python
from sklearn.pipeline import Pipeline

cv_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor_check),
        ("model", RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

print("Cross-validation pipeline created successfully!")
```

```python
from sklearn.model_selection import KFold, cross_val_score

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_scores = cross_val_score(
    cv_pipeline,
    X_check,
    y_check,
    cv=kf,
    scoring="r2",
    n_jobs=-1
)

print("5-Fold Cross-Validation R² Scores:")
print(cv_scores)

print("\nMean R²:", cv_scores.mean())
print("Standard Deviation:", cv_scores.std())
```

```python
# Train the final pipeline on the complete Bengaluru dataset

final_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor_check),
        ("model", RandomForestRegressor(
            n_estimators=100,
            random_state=42,
            n_jobs=-1
        ))
    ]
)

final_model.fit(X_check, y_check)

print("Final model trained successfully!")
```
